```python
# 구구단
# 8단
'''
8 * 1 = (8 * 1)
8 * 2 = (8 * 2)
8 * 3 = (8 * 3)
8 * 4 = (8 * 4)
8 * 5 = (8 * 5)
8 * 6 = (8 * 6)
8 * 7 = (8 * 7)
8 * 8 = (8 * 8)
8 * 9 = (8 * 9)
'''
start = 1
while start <= 9:
    print(f"8 * {start} = {start * 8}")
    #print("8 * %d = %d" % (start , 8 * start))
    #print("8 * {0} = {1}".format(start, start * 8))
    start += 1
```

```python
# break, continue
i = 1
sum1 = 0
while True:
    if i == 101:
        break # 반복문 종료
    sum1 = sum1 + i
    i += 1
print(sum1)
sum1 = 0
for i in range(1,10000000):
    if i == 101:
        break
    sum1 += i
print(sum1)
```

```python
# 1 ~ 100중 짝수만 더하시오
i = 1
sum1 = 0
while i <= 100:
    if i % 2 == 0:
        sum1 += i
    i += 1
print(sum1)
i = 0
sum1 = 0
while i <= 100:
    i += 1
    if i % 2 != 0:
        continue    
    sum1 += i
print(sum1)
```

```python
sum1 = 0
for i in range(1, 101):
    if i % 2 == 0:
        sum1 += i
print(sum1)
sum1 = 0
for i in range(1, 101):
    if i % 2 != 0:
        continue # 조건이 만족하면 for문으로 올라감
    sum1 += i
print(sum1)
```

```python
l = []
for dan in range(2, 10):
  for gop in range(1, 10):
    if dan * gop % 2 == 0:
      l.append(dan * gop)

l = [dan * gop for dan in range(2, 10) 
               for gop in range(1, 10) if dan * gop % 2 == 0 ]
```
