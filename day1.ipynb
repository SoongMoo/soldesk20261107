```python
# format 함수를 사용한 포매팅  : format()
# "apple": 문자열 , apple : 변수 , apple() : 함수, print():출력함수
a = "이숭무의 나이는 20살이고 키는 175.3입니다"
name = "이숭무"
age = 20
height = 175.3
a = "%s의 나이는 %d살이고 키는 %f입니다" % (name, age, height)
print(a)
a = "{1}의 나이는 {0}살이고 키는 {2}입니다".format(age, name,  height)
#                                                 0     1       2 
print(a)
a = "{1}의 나이는 {0}살이고 키는 {2}입니다".format(age, name,  173.5)
print(a)
```

```python
# format함수에 있는 이름 사용하기
a = "{1}의 나이는 {0}살이고 키는 {h}입니다".format(age, name, h = 173.5)
print(a)
a = "{1}의 나이는 {0}살이고 키는 {h}입니다".format(age, name, h = height)
print(a)
# a = "{n}의 나이는 {0}살이고 키는 {2}입니다".format(age, n = name, height)
#                                                 0      n        X
a = "{n}의 나이는 {0}살이고 키는 {h}입니다".format(age, n = name, h = height)
print(a)
```

```python
# 정렬
a = "{1}의 나이는 {0}살이고 키는 {h}입니다".format(age, name, h = height)
print(a)
a = "{1:10}의 나이는 {0:10}살이고 키는 {h:10}입니다".format(age, name, h = height)
print(a)
a = "{1:>10}의 나이는 {0:^10}살이고 키는 {h:<10}입니다".format(age, name, h = height)
print(a)
a = "{1:#>10}의 나이는 {0:$^10}살이고 키는 {h:=<10}입니다".format(age, name, h = height)
print(a)
```

```python
# 리스트 함수
### 리스트에 요소 추가
l = ['Life','is']
# 'too' 문자열 추가
l.append('too')
print(l)
l.append('short')
print(l)
l = ['Life','is']
# l.append('too', 'short') # TypeError: list.append() takes exactly one argument (2 given)
l.append(['too', 'short'])
print(l)
l = ['Life','is']
l.extend(['too', 'short'])
print(l)
l = ['Life','short']
l.insert(1, 'is')
print(l) # ['Life', 'is', 'short']
l.insert(2, 'too')
print(l)
# 데이터 추가 : append, extend, insert
```

```python
# 요소 삭제
l = ['Life', 'is', 'too', 'short']
#      0       1     2        3
# del 명령을 이용한 삭제 : index
del l[1] # indexing
print(l)
del l[1:1+2] # slicing
print(l)

#  요소를 이용한 삭제
l = ['Life', 'is', 'too', 'short']
l.remove("is")
print(l)
# l.remove('too', 'short') TypeError: list.remove() takes exactly one argument (2 given)
# 모두 삭제
l.clear() # l = []
print(l)
# 삭제한 데이터를 가져오기 : index를 사용
l = ['Life', 'is', 'too', 'short']
result = l.pop() # 마지막 요소 삭제한 값을 반환
print(l) 
print(result)
l = ['Life', 'is', 'too', 'short']
result = l.pop(2)
print(l)
print(result)
# del, remove, clear, pop
```
