```python
def calc(x1, x2, opt):
    if opt == "add":
        result = x1 + x2
    elif opt == "sub":
        result = x1 - x2
    elif opt == "mul":
        result = x1 * x2
    else:
        result = x1 / x2
    return result
```

```python
# 상속 : 이미 만들어져 있는 것을 가져와서 재사용
class Calculator: # 부모 클래스
    def __init__(self, first, second):
        self.first = first
        self.second = second
    def add(self):
        self.result = self.first + self.second
    def div(self):
        self.result = self.first / self.second
```

```python
class FourCalc(Calculator): # 자식 클래스
    # def __init__(self, first, second):
    #     self.first = first
    #     self.second = second
    # def add(self):
    #     self.result = self.first + self.second
    # def div(self):
    #     self.result = self.first / self.second
    def sub(self):
        self.result = self.first - self.second
    def mul(self):
        self.result = self.first * self.second
```

```python
a = FourCalc(100,200)
a.add()
print(a.result)
a.sub()
print(a.result)
```

```python
a = FourCalc(0,200) # 0 / 200 # 부정
a.div()
print(a.result)
```

```python
a = FourCalc(200,0) # 200 / 0 
# a.div() # ZeroDivisionError
# print(a.result) 
```

```python
class FourCalc(Calculator): # 자식 클래스
    def sub(self):
        self.result = self.first - self.second
    def mul(self):
        self.result = self.first * self.second
    def div(self): # 오버라이딩 overriding : 재졍의
        if self.second == 0:
            self.result = 0.0
        else:
            self.result = self.first / self.second
```

```python
a = FourCalc(200,0) 
a.div() 
print(a.result) 
```

```python
# 배열을 만드는 3가지 방법
# 1. random객체를 사용하는 방법
# 1차원 배열 : vector
num = np.random.randint(1, 10, 3) # 3차원 벡터
print(num)
num = np.random.randint(1, 10, 5) # 5차원 벡터
print(num)
num = np.random.randint(1, 10, size=5) # 5차원 벡터
print(num)
# 2차원 배열
num2 = np.random.randint(1, 10, (3, 2)) # 3행 2열
print(num2)
num2 = np.random.randint(1, 10, size=(2, 3)) # 2행 3열
print(num2)
l = [5, 5, 1]
print(l)
```

```python
# 2. arange()함수를 사용하는 방법
# 1차원 배열
num = np.arange(10) # 0 ~ 9
print(num)
num = np.arange(11, 21) # 11 ~ 20
print(num)
num = np.arange(11, 21, 2) # 11 ~ 20
print(num)
# 2차원 배열
num2 = np.arange(10).reshape(2,5) # 2행 5열
print(num2)
num2 = np.arange(10).reshape(5,2) # 5행 2열
print(num2)
num2 = np.arange(12).reshape(3,4) # 3행 4열
print(num2)

num2 = np.arange(1 , 1+3).reshape(1, 3) # 1행 3열
print(num2)
num2 = np.arange(1 , 1+3).reshape(3, 1) # 3행 1열
print(num2)

my_arr = np.arange(1000000)
print(my_arr)
my_arr = np.arange(1_000_000)
print(my_arr)
```

```python
#3. array()함수를 사용하는 방법: 리스트, 튜플
# 1 차원
num1 = np.array([10, 26, 15])
print(num1)
num1 = np.array((10, 26, 15))
print(num1)
l = [10, 26, 15]
num1 = np.array(l)
print(num1)
# 2 차원 배열
l2 = [[1,2,3,4],[5,6,7,8]]
num2 = np.array(l2)
print(num2)
num2 = np.array([[1,2,3,4],[5,6,7,8]])
print(num2)
# 누락값(null)을 갖는 배열
l1 = [10, np.nan, 20]
num1 = np.array(l1)
print(num1)
l2 = [[1,4,np.nan,5],[5,np.nan,8,4], [3,5,8,np.nan]]
num2 = np.array(l2)
print(num2)
```
